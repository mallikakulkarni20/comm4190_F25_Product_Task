```python
def query_LLM(messages=None, prompt=None, model='gpt-4o'):

    if messages is None and prompt:
        messages = [{ "role": "user", "content": prompt }]
    elif messages is None:
        return None
        
    completion = client.chat.completions.create(
      model=model,
      messages=messages
    )
    
    response = completion.choices[0].message.content

    return response
```

```python
system_prompt = "You are a consumer who is actively shopping for an all-weather technical jacket for commuting and travel. You care most about warmth, durability, comfortable fit, and style. You are deciding between two product websites. Judge them strictly as a shopper, not as a designer or marketer."

rating_prompt = f'''
   Read the websites for each of these two products
   1. JetJacket
   2. Levium

   {productA}

   ----

   {productB}

   ----
   
   Pay careful attention to the features, descriptions and user reviews.

   Rate each product on a scale of 0 (never buy) to 10 (buy immediately).
   Show just your ratings and which product you prefer.
   Then, briefly explain your ratings.
'''
```

```python
consumers = [
    "young professional, late 20s, travels frequently for work, values sleek minimalist style, prefers practical techwear, often works remotely from cafés or airports",
    "college student, early 20s, studies abroad often, budget-conscious but values durability, likes versatile clothing that transitions from class to travel, follows lifestyle influencers",
    "outdoor enthusiast, late 30s, enjoys weekend hiking and city exploration, prioritizes comfort and weather protection, shops for sustainable gear, tech-savvy and brand-aware",
    "retired traveler, mid 60s, enjoys cruises and guided tours, prioritizes comfort and easy maintenance, dislikes heavy luggage, appreciates functional pockets and simplicity"
]
```

```python
for consumer in consumers:
    print(f"\n-----\n{consumer}")

    messages = [
       {"role": "system", "content": system_prompt.format(description=consumer)},
       {"role": "system", "content": rating_prompt}
    ]

    response = query_LLM(messages=messages)

    print(response,'\n')
```
